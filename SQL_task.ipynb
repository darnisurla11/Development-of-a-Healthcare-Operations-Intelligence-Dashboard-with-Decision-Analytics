**SQL Interview Questions and Solutions**

```python
import sqlite3

conn = sqlite3.connect(':memory:')
cursor = conn.cursor()
cursor.execute("""
CREATE TABLE Employees (
    emp_id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    salary REAL,
    age INTEGER
)
""")

conn.commit()
print("Table created successfully!")
cursor.executemany("""
INSERT INTO Employees VALUES (?, ?, ?, ?, ?)
""", [
    (1, 'Ravi', 'IT', 50000, 25),
    (2, 'Priya', 'HR', 40000, 24),
    (3, 'Arun', 'IT', 60000, 28),
    (4, 'Sneha', 'Sales', 45000, 26),
    (5, 'Kiran', 'HR', 35000, 23)
])

conn.commit()
```

**Find Employees Earning More Than 45000**

Description: Use the WHERE clause to filter employees whose salary exceeds 45000.

```python
cursor.execute("""
SELECT * FROM Employees
WHERE salary > 45000
""")

print(cursor.fetchall())
```

**Sort Employees by Salary**

Description: Use ORDER BY to arrange employees from highest to lowest salary.

```python
cursor.execute("""
SELECT name, salary FROM Employees
ORDER BY salary DESC
""")

print(cursor.fetchall())
```

**Find the Highest Salary**

Description: Use the MAX() function to find the highest employee salary.

```python
cursor.execute("""
SELECT MAX(salary) FROM Employees
""")

print(cursor.fetchone())
```

**Calculate the Average Salary**

Description: Use the AVG() function to calculate the average salary of all employees.

```python
cursor.execute("""
SELECT AVG(salary) FROM Employees
""")

print(cursor.fetchone())
```

**Count Employees in Each Department**

Description: Use GROUP BY and COUNT() to count employees in each department.

```python
cursor.execute("""
SELECT department, COUNT(*)
FROM Employees
GROUP BY department
""")

print(cursor.fetchall())
```

**Find the Second-Highest Salary**

Description: Use a subquery and MAX() to find the second-highest distinct salary.

```python
cursor.execute("""
SELECT MAX(salary)
FROM Employees
WHERE salary < (
    SELECT MAX(salary)
    FROM Employees
)
""")

print(cursor.fetchone())
```

**Find Departments with More Than One Employee**

Description: Use GROUP BY and HAVING to display departments containing more than one employee.

```python
cursor.execute("""
SELECT department, COUNT(*)
FROM Employees
GROUP BY department
HAVING COUNT(*) > 1
""")

print(cursor.fetchall())
```

**Find Employees Whose Names Start with 'A'**

Description: The LIKE operator matches text patterns. The % wildcard represents any sequence of characters.

```python
cursor.execute("""
SELECT * FROM Employees
WHERE name LIKE 'A%'
""")
print(cursor.fetchall())
```

**Count the Total Number of Employees**

Description: The COUNT() function returns the total number of employee records.

```python
cursor.execute("""
SELECT COUNT(*) FROM Employees
""")
print(cursor.fetchone())
```

**Delete an Employee Record**

Description: The DELETE statement removes records that match a specified condition.

```python
cursor.execute("""
DELETE FROM Employees
WHERE emp_id = 5
""")
conn.commit()

print("Employee record deleted successfully")
```

**Add a New Column Using ALTER TABLE**

Description: The ALTER TABLE statement modifies the structure of an existing table. ADD COLUMN adds a new column.

```python
cursor.execute("""
ALTER TABLE Employees
ADD COLUMN city TEXT
""")
conn.commit()

print("New column added successfully")
```

**Display Unique Departments Using DISTINCT**

Description: The DISTINCT keyword retrieves only unique values, eliminating duplicates from the result.

```python
cursor.execute("""
SELECT DISTINCT department
FROM Employees
""")

print(cursor.fetchall())
```
