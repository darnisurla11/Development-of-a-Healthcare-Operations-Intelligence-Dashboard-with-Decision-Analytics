**Question 1: Reverse a String**

Write a function that takes a string as input and returns the string reversed. For example, reverse_string("hello") should return "olleh".

```python
def reverse_string(s):
    return s[::-1]
print(f"'hello' reversed is: {reverse_string('hello')}")
print(f"'world' reversed is: {reverse_string('world')}")
```

**Question 2: Check for Palindrome**

Write a function that checks if a given string is a palindrome. A palindrome reads the same forwards and backwards. For example, is_palindrome("madam") should return True.

```python
def is_palindrome(s):
    cleaned_s = ''.join(filter(str.isalnum, s)).lower()
    return cleaned_s == cleaned_s[::-1]
print(f"'malayalam' is palindrome: {is_palindrome('malayalam')}")
print(f"'darni' is palindrome: {is_palindrome('darni')}")
```

**Question 3: Find the First Non-Repeating Character**

Given a string, find the first non-repeating character in it and return its index. If it doesn't exist, return -1. For example, for s = "leetcode", return 0. For s = "loveleetcode", return 2.

```python
from collections import Counter

def first_non_repeating_char(s):
    char_counts = Counter(s)
    for i, char in enumerate(s):
        if char_counts[char] == 1:
            return i
    return -1
print(f"First non-repeating char index for 'loveleetcode': {first_non_repeating_char('loveleetcode')}")
print(f"First non-repeating char index for 'aabb': {first_non_repeating_char('aabb')}")
```

**Question 4: Find Largest Element in a List (Lists and Arrays)**

Write a Python function to find the largest element in a list of numbers. Assume the list is not empty.

```python
def find_largest_element(numbers):
    return max(numbers)
print(find_largest_element([11,4,5,2,88]))
print(find_largest_element([0, -5, -20]))
```

***Question 6:Find the Factorial of a Number**
Description: Multiply all integers from 1 to the given number.

```python
num = 5
fact = 1

for i in range(1, num + 1):
    fact *= i

print(fact)
```

**Question 7:Check Whether a Number Is Prime**

A prime number has exactly two factors: 1 and itself.

```python
num = 7
count = 0

for i in range(1, num + 1):
    if num % i == 0:
        count += 1

print("Prime" if count == 2 else "Not Prime")
```

**Question 8: Find the Sum of Digits**

Extract each digit using % 10 and add it to the sum.

```python
num = 1234
total = 0

while num > 0:
    total += num % 10
    num //= 10

print(total)
```
